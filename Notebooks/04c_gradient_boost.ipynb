```python
import sys; sys.path.append("../src/models")
from model_setup import *
import joblib
```

```python
hgb = HistGradientBoostingClassifier(
    categorical_features=list(range(len(CAT))), # first len(CAT) columns are categorical
    learning_rate=0.1,
    max_iter=1000, 
    max_leaf_nodes=31,
    min_samples_leaf=20,
    l2_regularization=0.0,
    early_stopping=True,
    class_weight="balanced",
    random_state=SEED
)

hgb_pipe = evaluate("hgb_baseline", hgb, prep=prep_hgb, notes="lr=0.1, leaves=31, balanced")
show_results()
```

```python
param_grid = {
    "clf__learning_rate": [0.03, 0.1],
    "clf__max_leaf_nodes": [15, 31, 63],
    "clf__min_samples_leaf": [20, 50, 100],
    "clf__l2_regularization": [0.0, 1.0, 10.0],
    "clf__class_weight": ["balanced", None],
}

hgb_search = GridSearchCV(
    Pipeline([("prep", prep_hgb), ("clf", hgb)]), param_grid=param_grid, cv=cv,
              scoring={"f1_macro": "f1_macro", "bal_acc": "balanced_accuracy", "ap_law": ap_law},
              refit="ap_law",  # pick model that ranks laws best and tune cutoff afterward
              return_train_score=True, verbose=2,
              )

hgb_search.fit(X_train, y_train, groups=groups_train)
```

```python
RESULTS = ROOT / "results"
RESULTS.mkdir(exist_ok=True)
pd.DataFrame(hgb_search.cv_results_).to_csv(RESULTS / "hgb_grid_cv_results.csv", index=False)
joblib.dump(hgb_search.best_params_, RESULTS / "hgb_grid_best_params.pkl")
hgb_search.best_params_
```

# out-of-fold diagnostics

```python
proba = cross_val_predict(hgb_tuned_pipe, X_train, y_train, groups=groups_train, cv=cv, method="predict_proba")
pred = CLASSES[proba.argmax(axis=1)]

print(classification_report(y_train, pre, labels=CLASSES, target_names=[SHORT[c] for c in CLASSES], zero_division=0))
for i, c in enumerate(CLASSES):
    ap, base = average_precision_score(y_train == c, proba[:, i]), (y_train==c).mean()
    print(f"AP {SHORT[c]:>6}: {ap:.3f} (random= {base:.3f}, lift = {ap / base:.1f}x)")
```

```python
from sklearn.metrics import f1_score
weight_grid = [0.5, 0.75, 1, 1.5, 2, 3, 4, 6, 8, 10]

best_f1, best_w = 0, (1, 1, 1)
for w_pass in weight_grid:
    for w_law in weight_grid:
        w = np.arrray([1, w_pass, w_law])
        pred_w = CLASSES[(proba * w).argmax(axis=1)]
        f1 = f1_score(y_train, pred_w, average="macro")
        if f1 > best_f1:
            best_f1, best_w = f1, tuple(w)

print(f"best weights (died, passed, law): {best_w} -> out of fold macro-F1 {best_f1:.3f}")
```
