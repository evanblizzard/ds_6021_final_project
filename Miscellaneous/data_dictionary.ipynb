NOMINATE (Nominal Three-step Estimate) is a statistical model to fit to roll-call votes. Places each legislator at a point on a policy map so that their votes are best explained by their distance from each bill's outcomes.
- Each legislator i has an ideal point x<sub>i</sub> in a low-dimensional space (usually 2-D)
- Each roll call j has two points: a Yea outcome z<sub>jy</sub> and a Nay outcome z<sub>jn</sub>
- A legislator votes for the outcome they prefer, which is the closest one to their ideal point. The line between the two outcome points and the "cutting line" that goes through it perpendicularly splits legislators into Yea and Nay voters.

There's a bunch more complicated math but that's basically it. 

The difference between nokken_pool and nominate is nominate fits across all Congresses at once while nokken_poole fits one Congress at a time. Nokken Poole will be better for capturing a member's movement over time. However, higher noise.

LES is a formula from bill counts: count each member's by bill type and stage. There are five stages: bills sponsored (BILL), action in committee (AIC), action beyond committee (ABC), passed the chamber (PASS), and became Law (LAW). Each has three types: commemorative (C), substantive (S), and substantive and significant (SS). The way substantiveness and significance has changed a bit in recent years so it will likely be best to use LES 1.0 for all of our models. The bill types are then weighted: commemorative bills get weight 1, substantive 5, and substantive and significant get 10. Each stage is turned into a chair: member's weighted count is then divided by weighted total across all members in that chamber and Congress (five shares, one per stage). The five shares are then added and rescaled (N/5 factor makes average LES = 1 in each Congress.)
